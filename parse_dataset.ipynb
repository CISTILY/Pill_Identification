# Parse dataset

```python
import os
import shutil

dataset_path = "D:/Pill_jpg_2025_Original"
output_path = "D:/Pill_jpg_2025_small"

pills_folder = (f for f in os.listdir(dataset_path))

for folderName in pills_folder:
    pill_path = os.path.join(dataset_path, folderName)
    pills_side = (f for f in os.listdir(pill_path))
    for side in pills_side:
        side_path = os.path.join(pill_path, side, "Mau")
        pill_image = (f for f in os.listdir(side_path))
        for image in pill_image:
            image_dir = os.path.join(side_path, image)
            modify_path = os.path.join(output_path, folderName)
            os.makedirs(modify_path, exist_ok=True)
            output_image_path = os.path.join(modify_path, image)
            shutil.copyfile(image_dir, output_image_path)
             
```
