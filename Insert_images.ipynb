```python
import os
import re

def rename_with_chain(folder, old_name, new_name):
    """
    Rename old_name -> new_name.
    If new_name exists, rename that file to next number (recursively).
    """

    def get_next_name(name):
        base, ext = os.path.splitext(name)
        # Find the last number in the base (e.g. IMG_E3365 -> 3365)
        match = re.search(r'(\d+)(?!.*\d)', base)
        if not match:
            raise ValueError(f"Filename {name} must contain a number")
        num_start, num_end = match.span(1)
        prefix = base[:num_start]
        number = int(match.group(1)) + 1
        return f"{prefix}{number}{ext}"

    def shift_file(current_name):
        """If file exists, shift it forward until free."""
        path = os.path.join(folder, current_name)
        if os.path.exists(path):
            new_name = get_next_name(current_name)
            shift_file(new_name)  # recursively push chain forward
            os.rename(path, os.path.join(folder, new_name))
            print(f"Renamed {current_name} -> {new_name}")

    # Start chain shift if target name already exists
    shift_file(new_name)

    # Finally, rename original file
    os.rename(os.path.join(folder, old_name), os.path.join(folder, new_name))
    print(f"Renamed {old_name} -> {new_name}")


# Example usage
folder_path = r"D:/Pill_Identification/First_batch_50_900/Source/JPG"
#rename_with_chain(folder_path, "IMG_E3969.JPG", "IMG_E3345.JPG")
rename_with_chain(folder_path, "IMG_E3970.JPG", "IMG_E3346.JPG")
rename_with_chain(folder_path, "IMG_E3971.JPG", "IMG_E3347.JPG")
rename_with_chain(folder_path, "IMG_E3972.JPG", "IMG_E3348.JPG")
rename_with_chain(folder_path, "IMG_E3973.JPG", "IMG_E3349.JPG")
rename_with_chain(folder_path, "IMG_E3974.JPG", "IMG_E3350.JPG")
```
